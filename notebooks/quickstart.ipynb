# AutoVision — Quickstart

The simplest possible workflow:

1. Pick a few categories you want to distinguish
2. `run()` scrapes images, trains a classifier, saves the best checkpoint
3. Call `predict()` on any image

No datasets, no labelling — just search terms.

> **Before you start:** install the project with `pip install -e ..` from the repo root, or run `import sys; sys.path.insert(0, '..')` in the first cell.

```python
import sys
sys.path.insert(0, "..")  # only needed if you haven't done `pip install -e ..`
```

## Train

One call — it handles scraping, splitting, training, and checkpointing.
Expect ~5 min on CPU or ~90 sec with a GPU.

```python
from autovision import run

model, classes = run(
    queries=["ferret", "cat", "dog"],
    n_images=100,   # images downloaded per class
    epochs=10,
)

print("Trained on:", classes)
```

## Predict

Point it at any image — the checkpoint from training is reloaded automatically.

```python
from autovision.trainer import predict

results = predict("path/to/your/image.jpg", top_k=3)

print(f"{'Class':<22} {'Confidence':>10}  Bar")
print("-" * 50)
for cls, conf in results:
    bar = "\u2588" * round(conf * 30)
    print(f"{cls:<22} {conf:>9.1%}  {bar}")
```
